```python
import pandas as pd
from sklearn.metrics import classification_report, matthews_corrcoef, roc_auc_score

# 1. Load your CSV file
df = pd.read_csv("test_predictions_with_uncertainty.csv")

# 2. Print overall performance on all test data
print("--- Overall Test Performance ---")
print(classification_report(df["Label"], df["PredLabel"]))
print("MCC Score:", matthews_corrcoef(df["Label"], df["PredLabel"]))
print("AUROC Score:", roc_auc_score(df["Label"], df["PredProb_class1"]))

# 3. Check Uncertainty Alignment
# Calculate average uncertainty for correct vs. incorrect predictions
df["Is_Correct"] = df["Label"] == df["PredLabel"]
print("\n--- Uncertainty Analysis ---")
print(df.groupby("Is_Correct")["Uncertainty"].mean())

# 4. Filter out high-uncertainty samples (keep only the 80% most confident predictions)
threshold = df["Uncertainty"].quantile(0.80)
confident_df = df[df["Uncertainty"] <= threshold]

print("\n--- Performance on Confident Samples (Top 80%) ---")
print(classification_report(confident_df["Label"], confident_df["PredLabel"]))
print("Confident MCC Score:", matthews_corrcoef(confident_df["Label"], confident_df["PredLabel"]))
```
