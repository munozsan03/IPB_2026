```python
import pandas as pd
import numpy as np
from sklearn import metrics

# Load your saved predictions
df = pd.read_csv('/raid/data/smunoz/catnip/EviCYP/model/results/data_splits/drug_bmfm-target_esmc/0/test_predictions_with_uncertainty.csv')

# Extract arrays
prob_arr = df['PredProb_class1'].values
gold_arr = df['Label'].values

# ================= TOP-K / RANKING METRICS (k = 10) =================
k = 10
sorted_indices = np.argsort(prob_arr)[::-1]
top_k_indices = sorted_indices[:k]
gold_top_k = gold_arr[top_k_indices]

precision_at_k = np.sum(gold_top_k) / k if k > 0 else 0.0
total_positives = np.sum(gold_arr)
recall_at_k = np.sum(gold_top_k) / total_positives if total_positives > 0 else 0.0

dcg = np.sum(gold_top_k / np.log2(np.arange(2, k + 2)))
ideal_top_k = np.sort(gold_arr)[::-1][:k]
idcg = np.sum(ideal_top_k / np.log2(np.arange(2, k + 2)))
ndcg_at_k = dcg / idcg if idcg > 0 else 0.0

global_pos_rate = total_positives / len(gold_arr) if len(gold_arr) > 0 else 0.0
enrichment_at_k = precision_at_k / global_pos_rate if global_pos_rate > 0 else 0.0

print(f"Precision@{k}: {precision_at_k:.4f}")
print(f"Recall@{k}: {recall_at_k:.4f}")
print(f"nDCG@{k}: {ndcg_at_k:.4f}")
print(f"Enrichment@{k}: {enrichment_at_k:.4f}")

```
